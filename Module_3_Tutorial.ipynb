Sort an Array of 0s, 1s, and 2s

```python
class Solution:
    def __init__(self, arr):
        # code here
        self.arr = arr

    def sort012(self):
        arr = self.arr
        new_arr = []
        num_zeros = 0
        num_ones = 0
        num_twos = 0
        if len(arr) > 0:
            for value in range(len(arr)):
                if arr[value] == 0:
                    num_zeros += 1
                elif arr[value] == 1:
                    num_ones += 1
                elif arr[value] == 2:
                    num_twos += 1
            for value in range(len(arr)):
                if num_zeros > 0:
                    new_arr.append(0)
                    num_zeros -= 1
                elif num_ones > 0:
                    new_arr.append(1)
                    num_ones -= 1
                elif num_twos > 0:
                    new_arr.append(2)
                    num_twos -= 1
        print('Original Array: {}\nSorted Array: {}'.format(arr,new_arr))
                
                    
                

solution1 = Solution([0,1,2,0,1,2])
solution2 = Solution([0, 1, 1, 0, 1, 2, 1, 2, 0, 0, 0, 1])
solution1.sort012()
solution2.sort012()
```

Binary Search

```python
class Solution:
    def __init__(self,arr,k):
        self.arr = arr
        self.k = k
        
    def firstSearch(self):
        # Code Here
        arr = self.arr
        k = self.k
        try:
            print(arr.index(k))
        except:
            print('Index is out of range!')

solution1 = Solution([1, 2, 3, 4, 5],4)
solution2 = Solution([11, 22, 33, 44, 55],445)
solution3 = Solution([1, 1, 1, 1, 2],1)
solution1.firstSearch()
solution2.firstSearch()
solution3.firstSearch()
        
```
